```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('species', axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn = KNeighborsClassifier(n_neighbors=5)
```

```python
model_knn.fit(X_train, y_train)
```

```python
model_knn.score(X_test, y_test)
```

```python
from sklearn.svm import SVC
```

```python
# model_svm = SVC(C = 10, kernel = 'linear', gamma = 'auto')
# we need to remove these for doing grid search cv the scores taken was of these all only
model_svm = SVC(gamma = 'auto')
```

```python
model_svm.fit(X_train, y_train)
```

```python
model_svm.score(X_test, y_test)
```

```python
#for this combo we r getting accuracy of 100% which means the model is overfit
# as we can see by changing values of hyperparameteres we are getting different values of scores to use the most efficient combination of the parameters now let's do grid search CV
```

# Grid Search CV

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV((model_svm), {
    'C' : [1,10,20,30],
    'kernel' : ['rbf','linear'],
},cv=5, return_train_score=False)
# here we are continously checking the cross validation at 5 to ensure our model will not get overfit i.e., getting accuracy of 100%
```

```python
# now for this we do not need to do the train test split we can use the x and y directly
classifier.fit(X, y)
```

```python
classifier.cv_results_
```

```python
results = pd.DataFrame(classifier.cv_results_)
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

# Random Search CV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifier_r = RandomizedSearchCV((model_svm), {
    'C' : [1,10,20,30],
    'kernel' : ['rbf','linear'],
},n_iter=4, cv=5, return_train_score=False)
```

```python
classifier_r.fit(X, y)
```

```python
results_r = pd.DataFrame(classifier_r.cv_results_)
results_r
```

```python
results_r[['param_C','param_kernel','mean_test_score']]
```
