# K-Means Clustering

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X, y_true = make_blobs(n_samples=500, centers=3, cluster_std=0.60, random_state=42)
X
```

```python
# let's make it clean so here it is
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])
```

```python
df
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
# as we can see here i only need x not y cause it is unsupervised and no role of output is here yet
```

# *Elbow Curve / Elbow Method

```python
# inertia = WCSS
inertia = []
K_range = range(1, 11)
```

```python
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
inertia
```

```python
# here we can clearly see our value of k as 3 also to cross checked we have taken data of 3 centers
# let's visualize it also
plt.plot(K_range, inertia, marker='o')
```

```python
kmeans_final = KMeans(n_clusters=3, random_state=42)
```

```python
cluster_labels = kmeans_final.fit_predict(X_scaled)
```

```python
cluster_labels
```

```python
df['cluster'] = cluster_labels
```

```python
sns.scatterplot(x=df['Feature_1'],
                y=df['Feature_2'],
                hue=df['cluster'],
                palette='viridis')
```

# *Another model to test K-means clustering

```python
from sklearn.datasets import make_moons
```

```python
X, y_true = make_moons(n_samples=500, noise=0.05, random_state=42)
```

```python
from sklearn.cluster import KMeans, DBSCAN
```

```python
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
df['cluster'] = kmeans_labels
```

```python
sns.scatterplot(x=df['Feature_1'],
                y=df['Feature_2'],
                hue=df['cluster'],
                palette='viridis')
#
```

```python
# this was exactly what the problem is non-circular data clustreing is like this so for that we have DBSCAN
```

# DBSCAN

```python
dbscan = DBSCAN(eps=0.3, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster'] = dbscan_labels
```

```python
sns.scatterplot(x=df['Feature_1'],
                y=df['Feature_2'],
                hue=df['dbscan_cluster'],
                palette='tab10')
```
