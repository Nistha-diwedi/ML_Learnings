# **Z- test**

```python
import numpy as np
from scipy.stats import norm
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172] #it'll be always greater than 30
pop_mean = 170
pop_std = 3
sample_mean = np.mean(sample)
n = len(sample)
```

```python
z_score = (sample_mean - pop_mean) / (pop_std / np.sqrt(n))
z_score
```

```python
p_value = 2 * (1-norm.cdf(abs(z_score)))
p_value
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject null hypothesis")
else:
    print("Accept null hypothesis")
```

# **T-test**

```python
import numpy as np
from scipy import stats
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
samp_mean = np.mean(sample)
samp_std = np.std(sample, ddof=1)
n = len(sample)
mean_population = 170
```

```python
t_stats = (samp_mean - mean_population) / (samp_std / np.sqrt(n))
t_stats
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t_stats), df=n-1))
p_value
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject null hypothesis")
else:
    print("Accept null hypothesis")
```

# **two-sample test**

```python
import numpy as np
from scipy import stats
```

```python
grp_a = [85,88,90,92,87,85,89,91,86,88]
grp_b = [82,84,80,83,81,79,78,85,84,83]
```

```python
t_stats, p_value = stats.ttest_ind(grp_a, grp_b, equal_var=False)
t_stats
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject null hypothesis")
else:
    print("Accept null hypothesis")
```
